```python
import pandas as pd

df = pd.read_csv(r'Dataset/diabetes_risk_prediction_dataset.csv')

df.sample(10)
```

```python
df.isna().sum()
```

```python
df.shape
```

```python
df.ffill(inplace=True)
df.sample(10)
```

```python
df.isna().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.dtypes
```

```python
df[['Family_History_Diabetes',
    'Hypertension',
    'Heart_Disease',
    'Fatty_Liver',
    'PCOS',
    'Doctor_Consultation_Needed']]
```

```python
mappings = {'Yes':True, 'No':False}
df['Family_History_Diabetes'] = df['Family_History_Diabetes'].map(mappings)
df['Hypertension'] = df['Hypertension'].map(mappings)
df['Heart_Disease'] = df['Heart_Disease'].map(mappings)
df['Fatty_Liver'] = df['Fatty_Liver'].map(mappings)
df['PCOS'] = df['PCOS'].map(mappings)
df['Doctor_Consultation_Needed'] = df['Doctor_Consultation_Needed'].map(mappings)
```

```python
df[['Physical_Activity_Level',
    'Diet_Quality',
    'Sugar_Intake_Level',
    'Stress_Level',
    'Alcohol_Consumption',
    'Medication_Adherence',
    'Diabetes_Risk']].value_counts()
```

```python
df['Diabetes_Risk'].value_counts()
```

```python
df = df.sample(n=10000,random_state=42)
df
```
